```python
from pathlib import Path
import sys, pandas as pd
ROOT=Path.cwd().resolve(); ROOT=ROOT.parent if ROOT.name=='notebooks' else ROOT
sys.path[:0]=[str(ROOT/'src'),str(ROOT/'chatbot_app'/'backend')]
from app.query import ConversationContext, plan_query
questions=['Tell me its fertilizer.','Can you give me the GrainWeight?','What does the model think the cultivar is?','What is Gsw, and what is its predicted value for this image?','Explain the difference between Cultivar and Fertilizer, then give both predictions.','Report Cultivar first, then Gsw, then GrainWeight.','List everything except Cultivar.','What are the three continuous measurements?','plz giv fertiliser n cultivar','wht r the predcted gsw and phips2']
rows=[]
for q in questions:
    p=plan_query(q)
    rows.append({'question':q,'intent':p.intent,'targets':', '.join(p.targets),'definition':p.include_definition,'prediction':p.include_prediction})
pd.DataFrame(rows)
```
