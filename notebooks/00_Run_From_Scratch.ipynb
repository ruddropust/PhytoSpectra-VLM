# PhytoSpectra-VLM: run from scratch
This notebook calls the clean `phytospectra` package. It does not contain hidden model logic and does not read `research_v2`. Every output is created under this repository's `artifacts/`, `results/`, and `reports/`.


```python
from pathlib import Path
import os, sys, subprocess
ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks': ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))
VENV_PYTHON = Path(r'E:\Habib\.ruddro\Scripts\python.exe')
RAW_DATA = Path(r'E:\Habib\PhytoChat\data\raw')
RUN_ID = 'final_3x5_v1'
SEED, FOLD = 2026, 0
if not VENV_PYTHON.is_file(): raise FileNotFoundError(VENV_PYTHON)
print({'root': str(ROOT), 'raw_data': str(RAW_DATA), 'pipeline_python': str(VENV_PYTHON), 'notebook_kernel': sys.executable})
```

## Frozen configuration and architecture audit


```python
from phytospectra.config import load_config
from phytospectra.model import PhytoCore
cfg = load_config(ROOT/'configs/experiment.yaml', root=ROOT, raw_data=RAW_DATA, run_id=RUN_ID, seed=SEED, fold=FOLD)
model = PhytoCore()
assert sum(p.numel() for p in model.parameters()) == 1_998_346
cfg.values['model'], sum(p.numel() for p in model.parameters())
```

## Execute one seed/fold
Set `RUN = True` only when you intend to train. Use `SMOKE = True` first.


```python
RUN = False
SMOKE = True
RESUME = True
if RUN:
    command = [str(VENV_PYTHON), '-m', 'phytospectra', 'all', '--root', str(ROOT), '--raw-data', str(RAW_DATA), '--run-id', RUN_ID, '--seed', str(SEED), '--fold', str(FOLD)]
    if SMOKE: command.append('--smoke')
    if RESUME: command.append('--resume')
    env = dict(os.environ); env['PYTHONPATH'] = str(ROOT / 'src')
    subprocess.run(command, cwd=ROOT, env=env, check=True)
else:
    print('Dry mode: change RUN to True to execute with the .ruddro environment.')
```
